```python
from datetime import datetime
# ========================================
# SISTEMA DE PRÉSTAMOS
# Versión 1 - Programación basica y uso de listas
# ========================================

# Lista de alumnos activos en el sistema
alumnos = [
          ["26040200", "Superman", "Ingeniería en Sistemas"],
          ["26040201", "A", "Ingeniería en Inteligencia Artificial"],
          ["26040202", "B", "Ingeniería en Sistemas"],
          ["26040203", "C", "Ingeniería en Inteligencia Artificial"],
          ["26040204", "D", "Ingeniería en Sistemas"]
          ]
# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras = [
                [ "PC-LC1-01", "LC1", "Python, VS Code", "Disponible" ],
                [ "PC-LC1-02", "LC1", "Python, Java, VS Code", "Prestada" ],
                [ "PC-LC1-03", "LC1", "AutoCAD, MATLAB", "Reservada" ],
                [ "PC-LC2-01", "LC2", "Python, VS Code", "Disponible" ],
                [ "PC-LC2-02", "LC2", "Python, Java, VS Code", "Prestada" ],
                [ "PC-LC2-03", "LC2", "AutoCAD, MATLAB", "Disponible" ],
                [ "PC-LC3-01", "LC3", "Python, VS Code", "Disponible" ],
                [ "PC-LC3-02", "LC3", "Python, Java, VS Code", "Prestada" ],
                [ "PC-LC3-03", "LC3", "AutoCAD, MATLAB", "Reservada" ],
                [ "PC-LC3-04", "LC3", "Python, VS Code", "Mantenimiento" ]
               ]
historial_prestamos = []
necesidades = []
conteo_programas = []

# 1. consultar
def consultar_computadoras():
  print("\n--- COMPUTADORAS ---")
  for i in range(len(computadoras)):
      print("Aula:", computadoras[i][1],"| ID:", computadoras[i][0] ,
            "| Software:", computadoras[i][2]," |Estado:", computadoras[i][3] )

# 2. buscar
def buscar_software():
  programa = input( "\nIngrese el software que desea buscar: ")
  print("\n--- RESULTADOS ---")
  encontrado = False
  for i in range(len(computadoras)):
    # la funcion (lower) convierte a minuscula el texto, de esta forma
    # nos aseguramos que el texto a buscar y la informacion de las listas
    # este en minuscula y no tengamos problema de que no encunentra un programa
    # porque este escrito de forma diferente usando mayusculas y minusculas
    if programa.lower() in computadoras[i].lower():
        print("ID:", computadoras[i][0], "| Aula:", computadoras[i][1],
              "| Software:", computadoras[i][2], "| Estado:", computadoras[i][3])
        encontrado = True

    if encontrado == False:
      print("No se encontraron computadoras con ese software.")

# 3. solicitar
def solicitar_computadora():
   print("\n===SOLICITAR PRESTAMO===")
   num_control = input("Numero de control del alumno")
   alumno_existe = False
   for i in range (len(alumnos)):
    if num_control == alumnos [i][0]:
      alumno_existe = True
      break
   if not alumno_existe:
      print ("Alumno no registrado")
      return
   programa = input("¿Qué programa necesita?")
   encontrado = False

   for i in range(len(computadoras)):
       if programa.lower() in computadoras[i][2].lower():
          encontrado = True
          if computadoras[i][3]=="Disponible":
              computadoras[i][3]="Prestada"
              print(f"Se asignó la computadora: {computadoras[i][0]} en el aula {computadoras[i][1]}")
              # guardar software, alumno, aula y id
              registro = {
                  "alumno": num_control,
                  "software": programa,
                  "aula": computadoras [i][1],
                  "ID" : computadoras [i][0]
              }
              historial_prestamos.append(registro)
              break
          else:
              print(f"Computadora {computadoras[i][3]} no disponible, estado: {computadoras[i][3]}")
       if not encontrado:
            print(f"No existen computadoras con el software")
            respuesta = input(f"¿Desea agregar '{programa}' a la lista de necesidades? (si/no): ")
            if respuesta.lower() == "si":
                fecha = datetime.now().strftime("%d/%m/%Y %H:%M:%S")
                necesidades.append({"programa": programa, "fecha": fecha})
                print("Agregado a necesidades con fecha:", fecha)
            else:
                print("No se agregó, puede ser error de escritura")
            return


# 4. devolver
def devolver_computadora():
  print("\n ===DEVOLVER COMPUTADORA====")
  id_pc = input("ID de la computadora")
  existe = False
  for i in range( len(computadoras) ):
    if computadoras[i][0] == id_pc:
      existe = True
      if computadoras[i][3]== "Prestada" or computadoras[i][3]=="Reservada":
        computadoras[i][3] = "Disponible"
        print("La computadora", id_pc, "ya esta disponible")
        break
      else:
        print("La computadora", id_pc, "no esta prestada/reservada")
  if existe == False:
    print("No existe la computadora", id_pc)

# 6. Agregar computadora
def agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  nueva_pc = [id_computadora, aula, programas, "Disponible"]
  computadoras.append(nueva_pc)
  print("\nComputadora agregada correctamente.")

# 7. Historial de préstamos
def historial():
  print("\n--- HISTORIAL DE PRESTAMOS DE COMPUTADORAS ---")
  if len(historial_prestamos) == 0:
    print ("Aún no hay préstamos registrados")
  else:
    for idx, p in enumerate(historial_prestamos, start=1):
      print(f"{idx} | Alumno: {p['alumno']} | Software: {p['software']} | Aula: {p['aula']} | ID: {p['ID']}")

# 8. Necesidades de software
def mostrar_necesidades():
    print("\n--- HISTORIAL DEE NECESIDADES ---")
    if len(necesidades) == 0:
        print("No hay necesidades registradas")
        return
    for i in range(len(necesidades)):
      print(f"{i+1} | Programa: {necesidades[i]['programa']} | Fecha: {necesidades[i]['fecha']} |")

# 9. Estadisticas
def estadisticas():
      # Conteo por carrera
  conteo_carreras = []
  for i in range(len(historial_prestamos)):
      num_control = historial_prestamos[i]['alumno']
      carrera = ""
    # Buscar la carrera en la lista de alumnos
      for j in range(len(alumnos)):
          if alumnos [j][0] == num_control:
              carrera = alumnos [j][2]
    # contar la carrera
      encontrado = False
      for k in range(len(conteo_carreras)):
          if conteo_carreras[k][0] == carrera:
              conteo_carreras [k][1] += 1
              encontrado = True
      if not encontrado:
        conteo_carreras.append([carrera, 1])
# --- Mostrar resultados ---
  print("\n\n=============================")
  print("\nEstadísticas por carrera:")
  for i in range(len(conteo_carreras)):
      print(conteo_carreras[i][0], ":", conteo_carreras[i][1], "veces")

#Estadisticas por programa
  print("\nEstadísticas por programas: ")
  conteo_programas = []
  for p in range(len(historial_prestamos)):
      programa = historial_prestamos [p]['software'].lower().strip()
      encontrado = False
      for k in range(len(conteo_programas)):
          if conteo_programas[k][0] == programa:
            conteo_programas[k][1] += 1
            encontrado = True
            break
      if not encontrado:
          conteo_programas.append([programa, 1])

  for i in range(len(conteo_programas)):
      print(conteo_programas[i][0], ":", conteo_programas[i][1])


# MENÚ PRINCIPAL
# Usamos un ciclo While que se repetira indefinidamente hasta que el usuario
# teclee la opcion 7 (salir) o cualquier numero superior a 7
opcion = 0
while opcion <= 10:
    print("\n================================")
    print("      SISTEMA DE PRÉSTAMOS")
    print("================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Mostrar historial de préstamos")
    print("8. Necesidades de software")
    print("9. Estadisticas")
    print("10. Salir")
    print("================================")
    opcion = int(input("Seleccione una opción: "))

    # 1. Consultar computadoras
    # si el usuario selecciono la opcion 1 se hace un ciclo (for) que recorre
    # toda la lista de computadoras e imprime los valores de las demás listas
    if opcion == 1:
      consultar_computadoras()

    elif opcion == 2:
        buscar_software()

    elif opcion==3:
        solicitar_computadora()

    elif opcion==4:
        devolver_computadora()

    elif opcion == 6:
        agregar_computadora()

    elif opcion == 7:
        historial()

    elif opcion == 8:
        mostrar_necesidades()

    elif opcion == 9:
      estadisticas()

    elif opcion == 10:
        print("\nGracias por utilizar el sistema.")
    else:
        print("\nOpción no válida.")


```
